```python
# Crea una nueva columna llamada 'title_length' con la longitud (número de caracteres) de cada título
# Importar librerias
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import skew, kurtosis, mode

# Cargar dataset de IMDb
url = "../assets/imdb_1000.csv"
df = pd.read_csv(url)
Longitud_title = df["title"].str.len()
df["Longitud_title"]=df["title"].str.len()
print(df.head(2))
```

```python
# Calcula media, mediana y moda de la columna 'title_length'
Media=df["Longitud_title"].mean()
Mediana=df["Longitud_title"].median()
Moda=df["Longitud_title"].mode()
print(f"Media: {Mediana}, la Mediana es: {Mediana} y finalmente la moda es: {Moda}")
df["Longitud_title"].describe()
```

```python
# Calcula el rango, la varianza y la desviación estándar de 'title_length'
range_ = df["Longitud_title"].max() - df["Longitud_title"].min()
Varianza=df["Longitud_title"].var()
Desviacion=df["Longitud_title"].std()
print(f"Rango: {range_}, Varianza: {Varianza}, y por ultimo la desviacion estandar es: {Desviacion}")

```

```python
# Calcula asimetría y curtosis de la columna 'title_length'
skewness = df["Longitud_title"].skew()
kurt = df["Longitud_title"].kurt()
print(f"Asimetría (Skewness): {skewness} y la Kurt es: {kurt}")
#es asimentrica posiva 
```

```python
# Imprime el título más corto y el título más largo según su longitud

indice_largo = df["Longitud_title"].idxmax()
indice_corto=df["Longitud_title"].idxmin()
titulo_mas_largo = df['title'].loc[indice_largo]
titulo_mas_corto=df["title"].loc[indice_corto]
print(f"El título más largo es: {titulo_mas_largo} y el titulo mas corto es: {titulo_mas_corto}")
```

```python
# grafica tus resultados
import matplotlib.pyplot as plt

df["Longitud_title"].hist(bins=5)
plt.title("Valores de los titulos")
plt.ylabel("Valores")
plt.xlabel("Frecuencia")
plt.show()
```

```python
import pandas as pd
import numpy as np
data = {'valores': [10, 20, 4, 5, 10, 5]}
df = pd.DataFrame(data)

media = df['valores'].mean()

df["diferencia"] = df["valores"] - media

df['diferencia_cuadrado'] = df['diferencia'] ** 2

print("DataFrame resultante (Primeras 2 filas):")
print(df.head(2))
```

```python


suma_cuadrados = df['diferencia_cuadrado'].sum()
n = len(df)

# Varianza muestral (dividir por n - 1 = 5)
varianza = suma_cuadrados / (n - 1)

# Desviación Estándar (raíz cuadrada)
desviacion_estandar = np.sqrt(varianza)

print(f"Suma Total de las Diferencias al Cuadrado: {suma_cuadrados:.2f}")
print(f"Número de observaciones (n): {n}")
print(f"VARIANZA: {varianza:.2f} (180.0 / 5)")
print(f"DESVIACIÓN ESTÁNDAR: {desviacion_estandar:.2f}")
```
